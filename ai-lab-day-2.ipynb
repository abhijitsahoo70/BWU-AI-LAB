Load the dataset and digits 

```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt 
digits = load_digits()
 
print("Data shape:", digits.data.shape)
print("Image shape:", digits.images.shape)
print("Target labels:", digits.target_names) 
plt.figure(figsize=(4, 4))
plt.imshow(digits.images[5], cmap='gray')
plt.title(f"Label: {digits.target[5]}")
plt.axis('off')
plt.show()
```

```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt 
digits = load_digits()

plt.figure(figsize=(10, 4)) 
for digit in range(10):
    index = list(digits.target).index(digit)

    plt.subplot(2, 5, digit + 1)
    plt.imshow(digits.images[index], cmap='gray')
    plt.title(f"Digit: {digit}")
    plt.axis('off')

plt.tight_layout()
plt.show()
```

1.Write a python program to import iris dataset from sk learn and convert it into dataframe and display shape of the dataset.


```python
 
from sklearn.datasets import load_iris
import pandas as pd 
iris = load_iris() 
df = pd.DataFrame(iris.data, columns=iris.feature_names) 
df["target"] = iris.target 
print(df)
 
print("\nShape of the dataset:", df.shape)
```

```python
 
from sklearn.datasets import load_iris, load_digits 
iris = load_iris()
digits = load_digits() 
iris_samples = iris.data.shape[0]
digits_samples = digits.data.shape[0] 
print("Number of samples in Iris dataset:", iris_samples)
print("Number of samples in Digits dataset:", digits_samples)
```
