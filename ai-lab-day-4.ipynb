BFS

```python
from collections import deque
import networkx as nx
import matplotlib.pyplot as plt

# Define the graph
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': [],
    'G': []
}

# BFS Function
def bfs(graph, start):
    visited = set()
    queue = deque([start])
    traversal = []

    visited.add(start)

    while queue:
        node = queue.popleft()
        traversal.append(node)

        for neighbour in graph[node]:
            if neighbour not in visited:
                visited.add(neighbour)
                queue.append(neighbour)

    return traversal

# Run BFS
result = bfs(graph, 'A')
print("BFS Traversal:")
print(" -> ".join(result))

# Create graph
G = nx.DiGraph()

for node in graph:
    for neighbour in graph[node]:
        G.add_edge(node, neighbour)

# Node positions
pos = {
    'A': (0, 3),
    'B': (-1, 2),
    'C': (1, 2),
    'D': (-1.5, 1),
    'E': (-0.5, 1),
    'F': (1, 1),
    'G': (0.5, 0)
}

# Draw graph
plt.figure(figsize=(8, 6))

nx.draw(
    G,
    pos,
    with_labels=True,
    node_color='skyblue',
    node_size=2500,
    font_size=12,
    font_weight='bold',
    arrows=True,
    edge_color='black',
    width=2
)

plt.title("BFS Graph")
plt.show()
```

DFS

```python
from collections import deque
import networkx as nx
import matplotlib.pyplot as plt

# Define the graph
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': [],
    'G': []
}

# DFS Function
def dfs(graph, start):
    visited = set()
    stack = [start]
    traversal = []

    while stack:
        node = stack.pop()

        if node not in visited:
            visited.add(node)
            traversal.append(node)

            # Reverse to visit left child first
            for neighbour in reversed(graph[node]):
                if neighbour not in visited:
                    stack.append(neighbour)

    return traversal

# Run DFS
result = dfs(graph, 'A')
print("DFS Traversal:")
print(" -> ".join(result))

# Create graph
G = nx.DiGraph()

for node in graph:
    for neighbour in graph[node]:
        G.add_edge(node, neighbour)

# Node positions
pos = {
    'A': (0, 3),
    'B': (-1, 2),
    'C': (1, 2),
    'D': (-1.5, 1),
    'E': (-0.5, 1),
    'F': (1, 1),
    'G': (0.5, 0)
}

# Draw graph
plt.figure(figsize=(8, 6))

nx.draw(
    G,
    pos,
    with_labels=True,
    node_color="skyblue",
    node_size=2500,
    font_size=12,
    font_weight="bold",
    arrows=True,
    edge_color="black"
)

plt.title("DFS Graph")
plt.show()
```
